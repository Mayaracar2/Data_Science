Ndarray (NumPY Array)

O python só aceita elementos do mesmo tipo.

SLICING NO NUMPY

```python
import numpy as np

#Plantando a mesma semente aleatória
np.random.seed(10)

mtz = np.random.randint(1, 99, 9).reshape(3,3)
print(mtz)
```

O slicing em ciência de dados, nos ajuda a picotar um dataset, que é considerado uma boa prática, ao envés de usar o dado todo.

```python
#Extraindo apenas a segunda linha da matriz
print(mtz[1])

#Extraindo apenas a terceira coluna da matriz
print(mtz[:,2])

#Extraindo a matriz 2x2 no canto inferior direito da matriz
#Os : serve para representar o ""Para frente"
print(mtz[1:,1:])
```

CONDICIONAIS NO NUMPY

O conceito de condicionais no numpy é o mais importante de todos, esse conceito mata todas as estruturas de repetição.

```python
print(mtz)

#Mostre apenas os elemntos menores que 70
print(mtz<70)
print(mtz[mtz<70])

#Retorne apenas os números pares
print(mtz%2==0)
print(mtz[mtz%2==0])
```

ANÁLISE DE PADRÕES TEXTUAIS COM NUMPY

```python
arr = np.array(['Inatel', 'Casa Viva', 'ICC', 'CDG', 'E-health', 'CSILab', 'ProdLab', 'CRA', 'CRR'])
print(arr)

#Submodulo do NumPy para trabalhar com textos: char
#Buscando qual texto aceita um padrão informado
arr = np.char.upper(arr)
print(np.char.find(arr, 'a'))
print((np.char.find(arr, 'a'))>=0)

cond = (np.char.find(arr, 'A'))>=0
print(arr[cond])
```

COMO IMPORTAR DATASETS NO NUMPY

```python
dataset = np.loadtxt('space.csv', delimiter=';', dtype='str')
print(dataset)
```

```python
#Extraindo as colunas do dataset
print(dataset[0,:])

#Extraindo o nome das empresas
print(np.unique(dataset[1:, 1], return_counts=True))
```
