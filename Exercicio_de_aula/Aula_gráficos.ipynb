FUNDAMENTOS DE SEABORN

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Importando o dataset tips
ds_tips = sns.load_dataset('tips')
print(ds_tips)

#Setando um estilo diferente no gráfico
sns.set_style('dark')

#Setando uma nova fonte
sns.set_context('talk')

#Traçando um scatterplot, scatter quer dizer dispersão
sns.scatterplot(data=ds_tips, x='total_bill', y='tip')
#Precisa de ajuda do matplotlib pois o seaborn é mais simples e objetivo
plt.xlabel('Conta com total em US$')
plt.ylabel('Gorjeta em US$')
plt.show()
```

HISTOGRAMA (Cria intervalos e vai distribuindo dentro desses intervalos)

A diferença do histograma e o do gráfico de barras, é que o histogram ele cria o intervalo e preenche sozinho.

```python
ds_penguins = sns.load_dataset('penguins')
print(ds_penguins.columns)

#Traçando o histograma
sns.histplot(
    data=ds_penguins, #Passando o dataset
    x='flipper_length_mm',
    hue= 'species',
    kde= 'True'
)
```

BOXPLOT

```python
ds_tips = sns.load_dataset('tips')

#Traçando um boxplot
sns.boxplot(
    data= ds_tips,
    x= 'day',
    y= 'tip',
    hue= 'sex'
)
plt.show()
```

REGPLOT

```python
sns.regplot(
    data=ds_tips,
    x= 'total_bill',
    y= 'tip',
    line_kws={'color':'red'}
)
plt.show()
```

HEATMAP

```python
#Selecionar as colunas que quero identificar uma correlação
corr = ds_tips[['total_bill', 'tip', 'size']].corr()

#Traçando o headmap
sns.heatmap(
    corr,
    annot= True,
    fmt= '.2f'
)
```
