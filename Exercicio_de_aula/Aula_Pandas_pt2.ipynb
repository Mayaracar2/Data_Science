CARREGANDO OS DATASETS COM PANDAS

```python
import numpy as np
import pandas as pd
```

```python
ds = pd.read_csv('paises.csv', sep=';')
print(ds)

#Visualizando as colunas
print(ds.columns)

#Extraindo registros do topo
print(ds.head(5))

#Extraindo registros da base
print(ds.tail(3))
```

SLICING NO DATASET PANDAS

```python
print(ds.columns)
print(ds[['Region', 'Climate',  'Service']])
```

CRIANDO NOVOS ARQUIVOS COM PANDAS

```python
#Calculando a porcentagem da população de cada país do mundo
total_polulacao = np.sum(ds['Population'])
print(total_polulacao)

#Calculando a porcentagem de cada país = broadcast
seriesPorcPais = (ds['Population']/total_polulacao)*100
print(seriesPorcPais)

#Adicionar esta Series no Dataset
ds['% Population'] = np.round(seriesPorcPais, 3)

#Criando uma v2 do dataset com coluna nova
ds.to_csv('paises_v2.csv')

#Pegando oos 5 países que mais tem gente e os 5 que tem menos gente
print(ds.nlargest(5, '% Population')['Country'])
print(ds.nsmallest(5, '% Population')['Country'])

```

AGRUPAMENTO DE DADOS NO PANDAS

OBS: SÓ FUNCIONA SE O DATASET TIVER AO MENOS UMA COLUNA DE DADOS CATEGÓRICOS

```python
#Agrupando por região
group_region = ds.groupby('Region')
print(group_region.count()['Country'])

#Somando a população de cada região
group_region = ds.groupby('Region')
print(group_region.sum()['Population'])
```

CRIAÇÃO E APLICAÇÕES DE FUNÇÕES CUSTOMIZDAS COM OS PANDAS

```python
#Criando uma função no python

#Função para dar 10% de desconto em alguma coisa
def tenPercent(x):
  return x * 0.9

```

```python
#Buscando a coluna que guarda a taxa de mortalidade de uma país
print(ds.columns)
taxa_mortalidade = ds['Deathrate']
print(taxa_mortalidade)

ds['Deathrate -10%'] = taxa_mortalidade.apply(tenPercent)

print(ds.head(2))
```
