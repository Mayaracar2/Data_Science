# 1. Tuplas
- guarda um conjunto de elementos
- é uma coleção imutável
- mas permite slicing dos seus elementos

```python
 nomes = ('Moranguinho', 'Limãozinha', 'Amoralinda', 'Laranjinha')
 print(nomes)
 print(type(nomes)) #questionando o Python para perguntar o tipo

 #tentando inserir um elemento na tupla
 #nomes[4] = 'Rocambole'#mas não é possivel, pois não permite. Não permite alterar os valores também

```

```python
#Slicing de Dados (fatiamento de dados)
print(nomes)
print(nomes[1])
print(nomes[1:3]) #o primeiro argumento do slicing é INCLUSIVE e o último argumento é EXCLUSIVE. Nesse acaso eu quera pegar apenas o Limãozinho e Amoralinda
print(nomes[1:])# da Limãozinha para frente
print(nomes[-2])# andando de trás para frente no vetor
```

# 2.Listas
- é uma coleção mutável
- mas permite slicing dos seus elementos

```python
nomes = ['Moranguinho', 'Limãozinha', 'Amoralinda', 'Laranjinha']
print(nomes)
print(type(nomes))

#CRUD(CREATE READ UPDATE DELETE) DE DADOS NA LISTA

##CREATE(duas formas de inserir na lista):
nomes.append('Harry Potter') #add no final
nomes.insert(2, 'Rony Wesley')
print(nomes)

##READ:
#mesmo procedimento que a tupla

#UPDATE
nomes[0] = 'Hermione'
print(nomes)

#DELETE
del nomes[1] #exclusao por indice
nomes.remove('Rony Wesley') #exclusao por valor
print(nomes)
```

# 3. Conjuntos
- não guarda elementos repetidos
- não guardam a ordem dos elementos
- permite naturalmente realizar operações entre conjuntos de dados(interseão, união, diferença)

```python
nomes = {'Moranguinho', 'Limãozinha', 'Amoralinda', 'Laranjinha'}
print(nomes)
print(type(nomes))

#CREATE
nomes.add('Barbie')
print(nomes)

#READ
print(nomes)
#print(nomes[0]) não fuciona pois não guarda a ordem
```

# 4. Dicionário
- organiza seus elementos no padrão chave:valor
- coleção mutável


```python
pessoa = {'nome': 'Spicy Girl',
          'idade' : 27,
          'sexo' : 'F'}
print(pessoa)
print(type(pessoa))

#CREATE
pessoa['Desenho'] = 'Angelina Bailarina'
print(pessoa)

#READ
print(pessoa['nome'])

#UPDATE
pessoa['Desenho'] = 'Backardigans'
print(pessoa)

#DELETE
del pessoa['sexo']
print(pessoa)
```

# 5.Misturando Coleções
- organiza seus elementos no padrão chave:valor
- coleção mutável


```python
#Dicionários
pessoa1 ={'nome' : 'Daniel', 'cidade' : 'Muzambinho'}
pessoa2 ={'nome' : 'Leonardo', 'cidade' : 'Alfenas'}
pessoa3 ={'nome' : 'Lara', 'cidade' : 'SRS'}

#Lista de dicionários
alunos = [pessoa1, pessoa2, pessoa3]

#Mostrando apenas os dados da pessoa2
print(alunos[1])

#Mostrando apenas a cidade da pessoa2
print(alunos[0]['cidade'])

#Mostrando dados da pessoa2 para frente
print(alunos[1:])
```
