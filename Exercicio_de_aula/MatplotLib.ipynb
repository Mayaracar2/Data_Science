#Cap.6 - MatplotLib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Plotando gráficos de linhas com MatplotLib (Plot)

```python
# eixo x
x = np.array([1, 2, 3, 4])
# eixo y
y = x*2

# eixo y2
y2 = x**2

# legendas nos eixos x e y
plt.xlabel('Valores X')
plt.ylabel('Valores Y')

# plotando o gráfico
plt.plot(x, y, 'o--r', x, y2, 's:b', linewidth=3, markersize=10)
plt.show()
print()

# Quando se cria um subplot, deve ser passado quantas linhas e colunas tem
plt.subplot(1, 3, 1)  #(linhas, colunas, index)
plt.title('Linear')
plt.plot(x, y, 'o--r', linewidth=3, markersize=10)
plt.subplot(1, 3, 2)
plt.title('Exponencial')
plt.plot(x, y2, 's:b', linewidth=3, markersize=10)
plt.show()
```

Scatter Plot - Gráfico de dispersão

```python
dfPaises = pd.read_csv('paises.csv', delimiter=';')

# extraindo as informações dos 6 maiores paises do mundo
dfMaiores = dfPaises.nlargest(6, 'Area (sq. mi.)')

# plotando qual destes paises possui maior renda per capita
plt.xlabel('Country')
plt.ylabel('GDP ($ per capita)')
# o parametro s permite adicionar uma terceira dimensao neste gráfico
plt.scatter(dfMaiores['Country'], dfMaiores['GDP ($ per capita)'], s=dfMaiores['Area (sq. mi.)']/10000)
plt.show()
```

Bar Plot - Gráfico em Barras

```python
# extraindo os 5 paises com maior PIB per capita (GDP) do dataset
dfBiggestGDP = dfPaises.nlargest(5, 'GDP ($ per capita)')

# pegando as Series dos paises e os valores de GDP separadas
dfBiggestGDP_country = dfBiggestGDP['Country']
dfBiggestGDP_gdp = dfBiggestGDP['GDP ($ per capita)']

# traçando um gráfico em barras para ilustrar as diferenças
plt.bar(dfBiggestGDP_country, dfBiggestGDP_gdp, color='green')
plt.show()
```

Pie Plot - Gráfico em pizza

```python
# extraindo os paises que não tem costa marinha
dfNoCoast = dfPaises[dfPaises['Coastline (coast/area ratio)'] == 0]

# extraindo a quantidade de paises que não tem costa marinha
qtdNoCoast = len(dfNoCoast)
# extraindo a quantidade de paises que tem costa marinha
qtdCoast = len(dfPaises) - qtdNoCoast

# traçando gráfico em torta
plt.pie(x=[qtdNoCoast, qtdCoast], labels=['% Países Sem Costa', '% Países Com Costa'], autopct='%1.1f%%')
plt.show()
```
