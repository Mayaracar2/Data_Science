PANDAS SERIES (1-D)


```python
import numpy as np
import pandas as pd
```

```python
#Como preencher uma Series
#Listas de labels
#Listas de valores

labels = ['Thiago', 'Matheus', 'Bruna', 'Julia']
valores = [23, 45, 34, 17]

#Criando a Series
se1 = pd.Series(index=labels, data=valores)
print(se1)
print(type(se1))

#Acessando os elementos de uma Series
print(se1['Matheus'])
print(se1[['Matheus','Julia']])
```

OPERAÇÕES ENTRE SERIES

```python
s1 = pd.Series({'v1': 10, 'v2': 45, 'v3':70})
s2 = pd.Series({'v1': 10, 'v4': 50, 'v3':70})

#print(s1)
#print(s2)

print(s1 + s2)
print(s1.add(s2, fill_value=0))

```

DATAFRAME (2-D+)

```python
#Como preencher um Dataframe
#Listas de labels (colunas)
colunas = ['w', 'x', 'y','z']

#Lista de labels (Linhas)
linhas = ['a', 'b', 'c', 'd', 'e']

np.random.seed(10)

#Listas de valores
valores = np.random.randint(1, 50, [5,4])

df = pd.DataFrame(columns=colunas, index=linhas, data=valores)

print(df)
```

MANIPULANDO O DATAFRAME

```python
print(df)

#Puxando uma unica coluna do dataframe
print(df['x'])
print(df[['y','z']])

#Puxando de uma única célula
print(df['y']['c'])

#Puxando multiplas colunas
print(df[['w','x','z']])

```

SLICING DE DADOS NO DATAFRAME COM LOC (LABELS) E ILOC (INDICES NUMERICOS)

```python
print(df)
```

```python
#Puxando uma única linha do dataframe
print(df.loc['c', ['w','x','y','z']])
print(df.iloc[2,:])

#Puxando multiplas linhas
print(df.loc[['b','e'], ['w', 'x', 'y', 'z']])
print(df.iloc[[1,4], :])

#Puxando a matriz 2x2 do canto inferior direito
print(df.loc[['d', 'e'], ['y', 'z']])
print(df.iloc[[3,4], [2,3]])
```
