CRIANDO UM NUMPY ARRAY






```python
import numpy as np

#CRIANDO UM NUMPY ARRAY 1D
arr = np.array([10,20,30,40,50,60])
print(arr)
print(type(arr))

#PROPRIEDADES DO ARRAY
print(arr.size) #quantidade de elementos
print(arr.ndim) #numero de dimensão
print(arr.shape) #formato do array

mtz = np.array(([10,20], [30,40], [50,60]))
print(mtz)
#PROPRIEDADES DO ARRAY
print(mtz.size)
print(mtz.ndim)
print(mtz.shape)
```

FUNÇÕES PRÉ-PRONTAS PARA ESTRUTURARMOS UM NUMPY ARRAYS

```python
#Ones
# cria um array uni ou bidimensional formado por 1`s
mtz = np.ones([5, 5])
print(mtz)

#zeros
# cria um array uni ou bidimensional formado por 0`s
arr = np.zeros(10)
print(arr.reshape(5,2)) #transforma um vetor em uma matriz

#Arange
# cria um array unidimensional formado por elementos espaçados
mtz = np.arange(2, 21, 2) #mostra a abrangência e o primeiro é  inclusivo e o segundo exclusivo
print(mtz.reshape(2,5))

```

OPERAÇÕES ENTRE NUMPY ARRAYS

```python
arr1 = np.array([10,20,30,40,50])
arr2 = np.array([60,40,20,10,5])
arr3 = arr1 + arr2

#Operações entre Arrays
print(arr3)
print(arr1 - arr2)
print(arr1 * arr2)

#Concatenação de Arrays
arr3 = np.concatenate([arr1, arr2])
print(arr3)

#Broadcasting - quando um ESCALAR faz uma operação com um ARRAY
print(5 * arr3)
```

```python
#ESTRUTURANDO UMA MATRIZ COM CONTAS
mtz = np.arange(10, 96, 5)
mtz = mtz.reshape(3, 6)
print(mtz)

#EXTRAINDO A MATRIZ DA PRIMEIRA COLUNA (NET)
print(mtz.sum(axis=0))
print(mtz.sum(axis=0)[0]) # Eixo 0 = coluna

print(mtz.sum(axis=1)) # Eixo 1 = linha
print(mtz.sum(axis=1)[1])
```

NÚMEROS ALEATÓRIOS NO NUMPY (MÓDULO RANDOM)

```python
#PLANTANDO A SEMENTE ALEATÓRIA
np.random.seed(5)
arr = np.random.randint(1, 10, 10)
print(arr)

#EXTRAINDO ELEMENTOS ÚNICOS
print(np.unique(arr))

#CONTANDO ELEMENTOS ÚNICOS
print(np.unique(arr, return_counts=True))
```
